```python
import numpy as np
import pandas as pd
df=pd.read_csv("Salaries.csv.zip")
```

```python
#total null value present in Data set
df.isna().sum().sum()
```

```python
# 1 display top 10 rows of the data set
df.head(10)
```

```python
# 2 display  last 10 rows of the data set
df.tail(10)
```

```python
# 3 find the shape of our data set (number of row & number of columns)
r,c=df.shape   # tuple unpacking
print("Number of Rows : ",r)
print("Number of columns : ",c)
```

```python
# 4 getting information about our data set like number of rows , total number of columns , data types of each columns and memory require
df.info()
```

```python
# 5 check the null value present in the data set
n=df.isna().sum().sum()
print("Total null values present in data set : ",n)
```

```python
# 6 drop ID ,Notes , Agency , and status columns
df=df.drop(columns=["Id","Agency","Notes","Status"])
```

```python
# 7 get statistics about data set
df.describe()      # include = "all"  that include numarical as well as all other columns
```

```python
# 8 Find the all occurance of the employee name (Top 5)
df["EmployeeName"].value_counts().head()
```

```python
# 9 find the number of unique job title

n=len(df["JobTitle"].unique())
print("Number of unique job title : ",n)
```

```python
# 10 total number of job title that contains "Captain"
df["JobTitle"].str.contains("CAPTAIN",case=False).sum()
# OR
df["JobTitle"].str.lower().str.contains("captain").sum()
```

```python
# 11 Display all employee names from fire deparment
df["EmployeeName"][df["JobTitle"].str.lower().str.contains("fire")]
```

```python
# 13 replace Not provide to "EmployeeName" to nan
df["EmployeeName"]=df["EmployeeName"].replace("Not Provided",np.nan)
```

```python
#14 Drop the rows having more than 5 missing values

df.drop(df[df.isnull().sum(axis=1)==5].index,axis=1,inplace=True)
```

```python
#15 find the job title of the "ALBERT PARDINI"
df["JobTitle"][df["EmployeeName"]=="ALBERT PARDINI"]
```

```python
#16 how much "ALBERT PARDINI" make (include benifits)
df[df["EmployeeName"]=="ALBERT PARDINI"]["TotalPayBenefits"]
```

```python
#17 display the name of person having highest Totalpay
df["EmployeeName"][df["TotalPay"]==df["TotalPay"].max()]
```

```python
# 18 find average total pay per year

df.groupby("Year")["TotalPay"].mean()
```

```python
#19 find the Average TotalPay of all employee per job title
df.groupby("JobTitle")["TotalPay"].mean()
```

```python
# 20 find average TotalPay of employee having job title ACCOUNTANT
df[df['JobTitle']=="ACCOUNTANT"]["TotalPay"].mean()
```
